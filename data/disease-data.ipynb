```python
import numpy as np
from scipy.stats import bernoulli

beta0 = -2
beta1 = 1
beta2 = 1.5


def sigma(X):
    return 1 / (1 + np.exp(-X))


def p(X):
    return sigma(beta0 + beta1 * X[0] + beta2 * X[1])


rng = np.random.default_rng(42)
X = bernoulli(1 / 2).rvs(size=(2, 5000), random_state=rng)
Y = bernoulli.rvs(p(X), random_state=rng)

data = np.vstack((X, Y)).T
np.savetxt("disease-data.csv", data, delimiter=",")
```
